**Print every word in upper case**

```python
words = ['play', 'filling', 'bar', 'theatre', 'easygoing', 'date', 'lead', 'that', 'story',  'island']
for i in words:
    print(i.upper())

```

**Create a new list containing only words with 5 or more letters**

```python
words = ['play', 'filling', 'bar', 'theatre', 'easygoing', 'date', 'lead', 'that', 'story', 'island']

lista_larga = []
for word in words:
    if len(word) >= 5:
        lista_larga.append(word)

print(lista_larga)

```

**Print the first word starting with "t"**

```python

for word in lista_larga:
    if word.startswith('t'):
        print(word)
        break
```

## Numbers

**Create a list containing the square of every number from 1 to 10**

```python
lista_square = []
for x in range(1, 11):
    lista_square.append(x**2)

print(lista_square)
```

**Print a list containing the square of every odd number from 1 to 10**

```python
lista_square = []
for x in range(1, 11):
    if x % 2 != 0:
        lista_square.append(x**2)

print(lista_square)
```

**Create a list with the squares of all multiples of 8 below 1000**

```python
lista_square = []
for x in range(8, 100, 8):
    if x :
        lista_square.append(x**2)

print(lista_square)
```

## People

```python
people = [
    {
        "name": "Juan",
        "age": 34,
        "n_kids": 2
    },
    {
        "name": "Pepe",
        "age": 27,
        "n_kids": 0
    },
    {
        "name": "Sonia",
        "age": 41,
        "n_kids": 1
    },
    {
        "name": "Lucía",
        "age": 22,
        "n_kids": 2
    },
    {
        "name": "Leo",
        "age": 55,
        "n_kids": 5
    }
]
```

**How many people are there?**

```python
print(len(people))
```

**How many people have kids**?

```python
cuanta_gente = 0

for persona in people:
    if persona["n_kids"] > 0:
        cuanta_gente += 1
```

**How many kids do they have in total?**

```python
total_niños = 0

for persona in people:
    total_niños += persona["n_kids"]

print(total_niños)
```

**In a year's time, names ending with "a" will have an extra kid. Create a list of dictionaries with people's info in a year's time**

```python
persona_nuevo_año = []

for persona in people:
    persona["age"] += 1

    if persona["name"].endswith("a"):
        persona["n_kids"] += 1
    
    persona_nuevo_año.append(persona)

print(persona_nuevo_año)
```
